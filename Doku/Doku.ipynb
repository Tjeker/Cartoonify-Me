## Ziel

Eine App, die Fotos in wählbaren Stilen cartoonisieren kann und dafür unterschiedliche Verfahren wie klassische Bildverarbeitung, GANs und einen Mehrfach-Adapter-Diffusionsansatz verwendet.

## Anforderungen

### Funktionale Anforderungen

- Nutzende können ein Bild im JPEG- oder PNG-Format mit einer Dateigröße von bis zu 10 MiB hochladen.
- Die Anwendung zeigt das ausgewählte Originalbild als Vorschau an. Nutzende können es entfernen oder durch ein anderes Bild ersetzen.
- Nutzende können zwischen mehreren visuell dargestellten Stilen wählen. TODO Welche
- Nutzende können das Ergebnisses beeinflussen, durch die Gewichtung von Konturen und Farbflächen.
- Nutzende können das erzeugte Bild herunterladen.
- Nutzende können sich bei CEWE myPhotos anmelden und ein Foto aus ihrem Konto zur Verarbeitung auswählen.

### Nichtfunktionale Anforderungen

- Das Hauptmotiv und die räumliche Anordnung des Originalbildes bleiben im Cartoonbild erkennbar.
- Gesichter werden nah am Original dargestellt.

## Persona

Jolanda, 23 möchte ein Persönliches Comic an ihre Beste Freundin verschenken mit Bildern die sie im Urlaub gemacht hat.

Jolanda hat in ihrem letzten Urlaub mit ihrer besten Freundin Julia viele Fotos mit ihrem Handy gemacht. Julia hat bald Geburtstag und Jolanda sucht noch nach einem Passenden geschenk. Julia ist ein großer Anime Fan deswegen will Jolanda eine Manga-Style Urlaubserinnerung erstellen.

### Typische Situation

Nach dem Urlaub setzt sich Jolanda abends an ihren Laptop und überlegt sich welche Fotos in das Buch kommen sollen.

Sie lädt das Bild hoch, probiert unterschiedliche Styles/Gewichtungen aus und vergleicht das Ergebnis mit dem Original. Beim ersten Versuch wirken die Konturen zu hart. Sie schiebt den Regler ein Stück Richtung Farbflächen und probiert es erneut.

### Was sie braucht

Aussagekräftige Stilbeispiele, einen Startwert für die Regler und einen direkten Vergleich mit dem Original. Begriffe wie „Semantik-Adapter“ helfen ihr bei der Entscheidung wenig. Sie benötigt eine möglichkeit die letzten einstellungen Profilbezogen zu speichern damit sie nicht immer nach den Korrekten einstellungen suchen muss.

### Wann sie zufrieden ist

Sie hat nach wenigen Versuchen eine Version gefunden, die wie Julias lieblings Anime aussieht und sie sofort an genau dieses Wochenende erinnert.

### Code zur Konturzeichnung

Unterschiede zwischen benachbarten Pixeln werden in beiden Bildrichtungen geprüft. Übersteigt mindestens ein Farbkanal den Kontrastschwellenwert 30, wird die Stelle schwarz markiert.

```python
edges = np.zeros(rgb.shape[:2], dtype=bool)
edges[:-1] |= np.any(np.abs(rgb[:-1] - rgb[1:]) > 30, axis=2)
edges[:, 1:] |= np.any(np.abs(rgb[:, 1:] - rgb[:, :-1]) > 30, axis=2)
contours = np.where(edges, 0, 255).astype(np.uint8)
Image.fromarray(contours).save("Papagei Konturen Beispiel.png")
```

### Code zur Schwarzweißdarstellung

Die Kantenmaske wird um alle Pixel ergänzt, deren Grauwert unter 100 liegt. So werden beispielsweise dunkle Teile des Schnabels und des Gefieders ausgefüllt. Beide Schwellenwerte stammen aus `black and white.py`.

```python
gray = np.asarray(image.convert("L"))
black_white = np.where(edges | (gray < 100), 0, 255).astype(np.uint8)
Image.fromarray(black_white).save("Papagei Schwarzweiss Beispiel.png")
```

### Code zur farbreduzierten Variante

Der Farbflächenprototyp im ZIP ist unvollständig. Dieses zusätzliche Beispiel zeigt eine mögliche Farbreduktion auf eine Palette mit 32 Farben. Es veranschaulicht den Effekt des gespeicherten Bildes, dessen vollständiger Erzeugungscode nicht vorliegt.

```python
reduced = image.quantize(colors=32).convert("RGB")
reduced.save("Papagei Farbflächen Beispiel.png")
```
